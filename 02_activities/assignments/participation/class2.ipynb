```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Create random data
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
fig.show()
```

```python
# Make our basic scatterplot:
fig, ax = plt.subplots(figsize=(5,3))

# Figure = set the entire canvas (in red) that contains everything
fig.set_facecolor('red') # changing the background color of the

# Axes = the plotting area inside a figure (the green boxes with x/y ticks where data isdrawn).
ax.set_facecolor('lightgreen') # changing the background color of the axes
ax.scatter(x, y)
fig.show()
```
